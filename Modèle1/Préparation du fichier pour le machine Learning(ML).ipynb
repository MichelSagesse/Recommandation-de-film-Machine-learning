# La méthode dummies

Diviser les genres

```python
# encodage + tranformation en booléen
data_genres = fusion_final["genres"].str.get_dummies(sep=',').astype(bool) 
data_genres
```

```python
# on concat sur les colonnes le data_genres avec fusion_final sans la colonne "genres"
data_diviser = pd.concat([fusion_final.drop(columns="genres"), data_genres ], axis=1)
data_diviser
```

Diviser les PrimaryName

```python
# encodage + tranformation en booléen
data_primaryName = fusion_final["primaryName"].str.get_dummies(sep=',').astype(bool) 
data_primaryName
```

```python
# on concat sur les colonnes le data_primaryName avec le data_diviser sans la colonne "primaryName"
data_diviser = pd.concat([data_diviser.drop(columns="primaryName"), data_primaryName ], axis=1)
data_diviser
```

Diviser les runtimeMinutes

```python
data_runtimeminutes= data_diviser["runtimeMinutes"].astype(str).str.get_dummies(sep=',').astype(bool) 
data_runtimeminutes
```

```python
# on concat sur les colonnes le data_primaryName avec le data_diviser sans la colonne "primaryName"
data_diviser = pd.concat([data_diviser.drop(columns="runtimeMinutes"), data_runtimeminutes ], axis=1)
data_diviser
```

# Diviser les Startyear

```python
# encodage + tranformation en booléen  en convertissant en string car get dummies est une methode en string
data_startyear = data_diviser["startYear"].astype(str).str.get_dummies(sep=',').astype(bool) 
data_startyear
```

```python
# on concat sur les colonnes le data_primaryName avec le data_dummies sans la colonne "startYear"
data_diviser = pd.concat([data_diviser, data_startyear ], axis=1)
data_diviser
```
