```python
# Fonction pour traiter chaque ligne du dataframe fusion_final1 et créer de nouvelles lignes
def expand_primary_name(row):
    if isinstance(row['primaryName'], str):
        primary_names = row['primaryName'].split(',')
        rows = []
        for name in primary_names:
            new_row = row.copy()
            new_row['primaryName'] = name
            rows.append(new_row)
        return pd.DataFrame(rows)
    else:
        return row

# Appliquer la fonction sur chaque ligne du dataframe  et concaténer les résultats
fusion_final1nettoye = pd.concat([expand_primary_name(row) for _, row in fusion_final1.iterrows()], ignore_index=True)
```
