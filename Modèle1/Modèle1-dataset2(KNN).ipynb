```python
X=data.iloc[:,8:-1]
X
```

```python
#Entrainer notre modèle: le nombre des k-voisins les plus proches est 5
model1=NearestNeighbors(n_neighbors = 5, metric = "cosine", algorithm = "brute").fit(X)
```

```python
#trouver le tconst d'un film,case permet au titre du film d'être sensible à la case
movie=data[data.primaryTitle.str.contains('X-Men',case=False)]
movie
```

```python
#tester avec un film
films = "X-Men (2000)"
data_film_choisi = data[ (data["primaryTitle"] == films) | (data["originalTitle"] == films) | (data["French_Title"] == films) ]
data_film_choisi
```

```python
film_choisi = data_film_choisi.iloc[:, 8:-1]
film_choisi
```

```python
#Les cinq voisins les plus proches du film choisi : distance et indices
distance, indice = model1.kneighbors(film_choisi)
```

```python
distance, indice
```

```python
pd.DataFrame(indice)
```

```python
indices=indice[0,1:]
indices
```

```python
# On affiche les 4 plus proches voisins 
print("On peut remplacer", films, "par :", data.iloc[indice[0,1:]]['primaryTitle'].values) 
```

```python
# le dataframe des suggestions 
suggestion = data.iloc[indice[0,1:]]
suggestion
```

```python
# Evaluation du modèle 
from sklearn.metrics.pairwise import cosine_similarity

# Calculer la similarité cosinus entre le film choisi et ses voisins recommandés
similarities = cosine_similarity(film_choisi, X.iloc[indices])

# Afficher les similarités pour chaque voisin recommandé
for i, sim in enumerate(similarities[0]):
    print(f"Similarité avec le voisin {i + 1}: {sim}")
```
