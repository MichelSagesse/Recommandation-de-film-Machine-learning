# Liste des films

```python
# colonne "primarytitle" en  DF
primarytitle = data_machine_learning[['tconst', 'primaryTitle']]
primarytitle.head()
```

# Listes des genres

```python
fusion_final1 = pd.read_csv(r"C:\Users\user\Downloads\Projet ML\Model1(KNN)\Données nettoyées\fusion_final1.csv")
fusion_final1
```

```python
#récupération des données de la colonne "genres" en un Dataframe
liste_genres = fusion_final1.groupby(["genres"]).count().reset_index().sort_values(by = "tconst", ascending = False).iloc[:,:2]
liste_genres
```

```python
# get dummies pour séparer les genres
liste_genres = liste_genres["genres"].str.get_dummies(sep=',').astype(bool)
liste_genres
```

```python
# récupérer le titre de chaque colonne 
columns = liste_genres.columns
columns
```

```python
# On stocke les genres dans un nouveau dataframe
liste_genres = pd.DataFrame({"genres" : columns})
liste_genres
```

```python
#Enregistrer les genres
liste_genres.to_pickle(r"C:\Users\user\Downloads\Projet ML\Model1(KNN)\Données nettoyées\liste_genres.pkl.gz", compression = "gzip")
```
