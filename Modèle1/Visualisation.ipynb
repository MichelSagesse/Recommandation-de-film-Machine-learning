```python
import pandas as pd
import plotly.express as px
```

```python
fusion_titlebasic_titlerating =pd.read_csv(r"C:\Users\user\Downloads\Projet ML\Model1(KNN)\Données nettoyées\fusion_titlebasic_titlerating.csv", sep = ",")
data_title_principal_nettoye = pd.read_csv(r"C:\Users\user\Downloads\Projet ML\Model1(KNN)\Données nettoyées\data_title_principal_nettoye.csv", sep = ",")
data_name_basics_purify = pd.read_csv(r"C:\Users\user\Downloads\Projet ML\Model1(KNN)\Données nettoyées\data_name_basics_purify.csv", sep = ",")
```

```python
fusion_namebasic_titleprincipal = pd.merge(data_title_principal_nettoye, data_name_basics_purify, 
         how="left",     
         left_on='nconst',
         right_on='nconst')
```

```python
viz=pd.merge(fusion_titlebasic_titlerating, fusion_namebasic_titleprincipal, 
         how="left",     
         left_on='tconst',
         right_on='tconst')
```

```python
viz
```

```python
viz=viz.drop(['titleType','Unnamed: 0','characters'],axis=1)
viz
```

```python
viz.to_csv(r"C:\Users\user\Downloads\Projet ML\Visualisation.csv")
```

```python
#Les cinq genres les plus representés
top5_genres = viz.groupby(['genres']).count().reset_index().sort_values(by = "tconst" , ascending = False)[:5]
top5_genres
```

```python
fig= px.pie(top5_genres, values="tconst", names = "genres", title = "Top 5 genres", hover_data=["genres"], labels = {"tconst" : " Total"} )
fig.show()
```

```python
#Nombre de film par an
data_nblignes_per_year = viz.groupby(by="startYear").count().reset_index()
data_nblignes_per_year
```

```python
data_nb_films_per_year = viz.groupby(["startYear","tconst"]).count().reset_index()

data_nb_films_per_year_1 = data_nb_films_per_year.groupby(["startYear"]).count().reset_index()
```

```python
data_nb_films_per_year
```

```python
fig1 = px.bar(data_nb_films_per_year_1, x = "startYear", y= "primaryTitle", title = "Nombre de films par an" )
fig1.update_xaxes(title="Années") 
fig1.update_yaxes(title="Nombre de films")
fig1.show()
```
